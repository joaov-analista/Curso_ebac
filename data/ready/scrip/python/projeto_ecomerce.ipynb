```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv("C:\\Users\\jv001\\Downloads\\GitHub\\curso_ebac\\data\\raw\\ecommerce_estatistica.csv")
print(df.head())
# Verifique a quantidade de linhas e colunas
linhas_colunas = df.shape
print("Verificar a qtd de Linhas e colunas: ", linhas_colunas)
```

```python
# Verifique os tipos de dados
tipos = df.dtypes
print('Verificar Tipagem:\n', tipos)
```

```python
# Verifique a quantidade de valores nulos
nulos = df.isnull().sum()
print('Verificar valores nulos:\n', nulos)
```

```python
#  Substitua os valores nulos das colunas 
# ‘Temporada’ e ‘Marca’ por ‘Não Definido’



df.fillna({'Temporada': 'Não Definido'}, inplace=True)
df.fillna({'Marca': 'Não Definido'}, inplace=True)
```

```python
# Converter a coluna 'Marca' Material Temporada para letras minúsculas

df['Marca'] = df['Marca'].str.lower()
df['Material'] = df['Marca'].str.lower()
df['Temporada'] = df['Temporada'].str.lower()


# Remover linhas duplicadas
print("QTD linhas duplicadas: ", df.shape[0])
df.drop_duplicates(inplace=True)
print("QTD linhas depois de dulplicadas: ", len(df))

# Remover linhas com menos de 8 valores não nulos
# O parâmetro 'thresh' define o número mínimo de valores 
# não nulos necessários para manter a linha
df.dropna(thresh=8, inplace=True)



```

```python
# Calcular o intervalo interquartil (IQR)
q1 = df['N_Avaliações'].quantile(0.25)
q3 = df['N_Avaliações'].quantile(0.75)
iqr = q3 - q1

# Definir o limite superior para identificar outliers
limite_alto = q3 + 1.5 * iqr

# Filtrar os produtos que possuem um número de avaliações 
# maior que o limite superior
df_avaliados = df[(df['N_Avaliações'] > limite_alto)]
```

```python

df['Desconto'].astype(str)

# Extrair e limpar os dados da coluna 'Condicao'
# A função lambda é usada aqui para pegar a primeira 
# palavra da string na coluna 'Condicao'
# x.split(' ')[0] pega a primeira palavra da string.
df['Condicao_Atual'] = df['Condicao'].apply(lambda x: x.split(' | ')[0])

# A função lambda é usada aqui para pegar a quinta palavra 
# da string na coluna 'Condicao' se existir,
# caso contrário, retorna 'Nenhum'
df['Qtd_Vendidos'] = df['Condicao'].apply(lambda x: x.split(' | ')[1] if len(x.split(' | ')) > 1 else 'Nenhum')


# A função lambda é usada aqui para remover o símbolo '%' 
# da string na coluna 'Desconto'
df['Desconto'] = df['Desconto'].apply(lambda x: x.replace('%', '') if isinstance(x, str) else x)
```
